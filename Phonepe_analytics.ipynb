```python
#Most frequent (dominant) transaction type per state & quarter
grouped = state_txn_split.groupby(['State', 'Year', 'Quarter', 'Transaction Type'])['Transactions'].sum().reset_index()
dominant_type = grouped.loc[grouped.groupby(['State', 'Year', 'Quarter'])['Transactions'].idxmax()]
dominant_type = dominant_type.reset_index(drop=True)
print(dominant_type.head(10))
```

```python
# Merge district transaction data with demographics on State + District
district_merged = district_txn.merge(
    District_Demographics,
    on=['State', 'District'],
    how='left'
)
district_merged['Users_to_Population'] = district_merged['Registered Users'] / district_merged['Population']
top_populated = district_merged[district_merged['Year'] == 2021].sort_values('Population', ascending=False).head(10)
print(top_populated[['State', 'District', 'Population', 'Registered Users', 'Users_to_Population']])
```

```python
# Keep only the last quarter for a clean, deduplicated year-end snapshot
year_end = district_merged[(district_merged['Year'] == 2021) & (district_merged['Quarter'] == 2)]

top_populated = year_end.sort_values('Population', ascending=False).head(10)
print(top_populated[['State', 'District', 'Population', 'Registered Users', 'Users_to_Population']])
```

```python
# Sum registered users by brand for each State-Year-Quarter (in case of duplicates)
brand_grouped = State_DeviceData.groupby(['State', 'Year', 'Quarter', 'Brand'])['Registered Users'].sum().reset_index()
```

```python
# Find the dominant brand per State-Year-Quarter using idxmax
dominant_brand = brand_grouped.loc[brand_grouped.groupby(['State', 'Year', 'Quarter'])['Registered Users'].idxmax()]
dominant_brand = dominant_brand.reset_index(drop=True)

print(dominant_brand.head(10))
```

```python
#all time dominant brand
overall_brand = State_DeviceData.groupby(['State', 'Brand'])['Registered Users'].sum().reset_index()
overall_dominant = overall_brand.loc[overall_brand.groupby('State')['Registered Users'].idxmax()]
overall_dominant = overall_dominant.reset_index(drop=True)

print(overall_dominant.sort_values('Registered Users', ascending=False).head(10))
```

```python
# Year-over-year growth % per state
yearly_totals = state_txn.groupby(['State', 'Year'])['Transactions'].sum().reset_index()
yearly_totals = yearly_totals.sort_values(['State', 'Year']) 

yearly_totals['YoY_Growth_%'] = yearly_totals.groupby('State')['Transactions'].pct_change() * 100
print(yearly_totals.head(10))
```

```python
yearly_totals.dropna().sort_values('YoY_Growth_%', ascending=False).head(10)  # biggest growth
```

```python
yearly_totals.dropna().sort_values('YoY_Growth_%').head(10)                   # biggest declines
```

```python
# Top and lowest states by total transactions, amount, and registered users (all-time)
top_states= state_txn.groupby('State')[['Transactions','Amount (INR)','Registered Users']].sum()
top_states=top_states.sort_values('Transactions',ascending=False)

print("Top 10 states by Transactions: ")
print(top_states.head(10)) 

print("\n Bottom 10 states by transactions: ")
print(top_states.tail(10))
```

```python
# District-level totals (latest year, Q4 snapshot to avoid duplicate rows)
district_latest = district_merged[(district_merged['Year'] == 2021) & (district_merged['Quarter'] == 2)]
top_districts=district_latest.sort_values('Transactions',ascending=False).head(10)
print("Top 10 districts by transactions:")
print(top_districts[['State', 'District', 'Transactions', 'Registered Users', 'Population']])

# Low adoption = low Users_to_Population ratio, but only among reasonably populated districts
low_adoption = district_latest[district_latest['Population'] > 100000].sort_values('Users_to_Population').head(10)
print("\nLow adoption districts (population > 100k):")
print(low_adoption[['State', 'District', 'Population', 'Registered Users', 'Users_to_Population']])
```

```python
# Correlation between Density/Population and Transactions/Registered Users
corr_cols = district_latest[['Population', 'Density', 'Transactions', 'Registered Users', 'Users_to_Population']]

correlation_matrix = corr_cols.corr()
print(correlation_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
plt.title('Correlation: Demographics vs Transaction Activity')
plt.show()
```

```python
# App opens trend over time, per state
engagement = state_txn.groupby(['State', 'Year'])['App Opens'].sum().reset_index()

# Top 5 states by total app opens (all years)
top_engaged_states = engagement.groupby('State')['App Opens'].sum().sort_values(ascending=False).head(5).index

engagement_top5 = engagement[engagement['State'].isin(top_engaged_states)]

plt.figure(figsize=(9, 5))
sns.lineplot(data=engagement_top5, x='Year', y='App Opens', hue='State', marker='o')
plt.title('App Opens Trend — Top 5 Most Engaged States')
plt.ylabel('Total App Opens')
plt.show()
```

```python
engagement_ratio = state_txn.groupby('State')[['App Opens', 'Registered Users']].sum()
engagement_ratio['Opens_per_User'] = engagement_ratio['App Opens'] / engagement_ratio['Registered Users']

print(engagement_ratio.sort_values('Opens_per_User', ascending=False).head(10))
```
